```python
# Calculate z-scores of percent changes of demographic variables 

from scipy import stats
import pandas as pd

# Read time series csv
neighborhood_years = pd.read_csv("data/denver_neighborhood_gentrification_timeseries.csv")
```

```python
denver_2011 = neighborhood_years[neighborhood_years['year'] == 2011]
denver_2022 = neighborhood_years[neighborhood_years['year'] == 2022]

denver_2011['pct_white_nonhispanic'] = denver_2011['white_nonhispanic'] / (denver_2011['white_nonhispanic'] + denver_2011['black_nonhispanic'] + denver_2011['hispanic_latino'])
denver_2022['pct_white_nonhispanic'] = denver_2022['white_nonhispanic'] / (denver_2022['white_nonhispanic'] + denver_2022['black_nonhispanic'] + denver_2022['hispanic_latino'])

denver_2011
```

```python
denver_2022
```

```python
# Calculate percent changes

denver_2011_idx = denver_2011.set_index('neighborhood')
denver_2022_idx = denver_2022.set_index('neighborhood')

denver_changes = denver_2022_idx.copy()

denver_changes['rent_growth_pct'] = (
    (denver_2022_idx['real_median_gross_rent'] - denver_2011_idx['real_median_gross_rent'])
    / denver_2011_idx['real_median_gross_rent'] * 100
)
denver_changes['home_value_growth_pct'] = (
    (denver_2022_idx['real_median_home_value'] - denver_2011_idx['real_median_home_value'])
    / denver_2011_idx['real_median_home_value'] * 100
)
denver_changes['household_income_growth_pct'] = (
    (denver_2022_idx['real_median_household_income'] - denver_2011_idx['real_median_household_income'])
    / denver_2011_idx['real_median_household_income'] * 100
)
denver_changes['white_nonhispanic_growth_pct'] = (
    (denver_2022_idx['pct_white_nonhispanic'] - denver_2011_idx['pct_white_nonhispanic'])
    / denver_2011_idx['pct_white_nonhispanic'] * 100
)

denver_changes = denver_changes.reset_index()  # bring 'neighborhood' back as a column
denver_changes = denver_changes.drop(columns=['year', 'total_pop', 'white_nonhispanic', 'black_nonhispanic', 'hispanic_latino', 'real_median_household_income', 'real_median_gross_rent',  'real_median_home_value', 'boundary_vintage', 'pct_white_nonhispanic'])

denver_changes

```

```python
from scipy import stats

# summary: one row per neighborhood, columns are your *change* metrics
# (e.g., % rent growth 2011->latest, % income growth, pct-point change in white_nonhispanic_share)

z_cols = []
for col in ["rent_growth_pct", "household_income_growth_pct", "white_nonhispanic_growth_pct"]:
    z_col = f"{col}_z"
    denver_changes[z_col] = stats.zscore(denver_changes[col], nan_policy="omit")
    z_cols.append(z_col)

denver_changes["gentrification_pressure_index"] = denver_changes[z_cols].mean(axis=1)

denver_changes
```

```python
denver_changes.to_csv("data/denver_changes.csv")
```
